```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def differenza(x):
    return np.sqrt(x+1)-np.sqrt(x)
```

```python
vals = np.logspace(13.5,17,1000)
y = differenza(vals)
```

```python
fig, ax = plt.subplots(figsize=(12,6))
ax.set_xscale('log')
ax.plot(vals,y)                       
```

```python
def differenza_stabile(x):
    return 1./(np.sqrt(x+1)+np.sqrt(x))
```

```python
vals1 = np.logspace(13.5,17,1000)
y1 = differenza_stabile(vals1)
```

```python
fig1, ax1 = plt.subplots(figsize=(12,6))
ax1.set_xscale('log')
ax1.plot(vals1,y1)                       
```

```python
def differenza_relativa(x):
    return (np.sqrt(x+1)-np.sqrt(x))/(np.sqrt(x+1)+np.sqrt(x))
```

```python
vals2 = np.logspace(13.5,17,50)
y2 = differenza_relativa(vals2)
```

```python
for x,y in zip(vals2,y2):
    print(f"{x:10.4e},{y:10.4e}")
```
