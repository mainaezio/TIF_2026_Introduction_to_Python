Scientific Python fornisce diverse routines di integrazione. Uno strumento di uso generale per calcolare integrali *I* del tipo

$$I=\int_a^b f(x) \mathrm{d} x$$

&egrave; la funzione `quad()` del modulo `scipy.integrate`.

Prende in input la funzione *f(x)* da integrare (l'“integrando”) e gli estremi inferiore e superiore *a* and *b*. 
Restituisce due valori, (in una tuple): il primo &egrave; il risultato dell'integrale mentre il secondo &egrave; una stima dell'errore numerico del risultato.

````python
from scipy import integrate
help(integrate.quad)
````

Ecco un esempio:
$$I=\int_{-2}^{2} e^{\cos(-2 \pi x)} \,\mathrm{d} x$$

````python
from math import cos, exp, pi
from scipy.integrate import quad

# funzione da integrare
def f(x):
    return exp(cos(-2 * x * pi))

# chiamata a quad
res, err = quad(f, -2, 2)

print(f"The numerical result is {res:f} (+-{err:g})")
````

Usando Numpy:

````python
import numpy as np

def f_N(x):
    return np.exp(np.cos(-2 * x * np.pi))


res, err = quad(f_N, -2, 2)
print(f"The numerical result is {res:f} (+-{err:g})")
````

Si noti che `quad()` pu&ograve; prendere come parametri opzionali `epsabs` e `epsrel` per aumentare o diminuire l'accuratezza del calcolo (Usate `help(quad)` per maggiori informazioni). I valori di default sono `epsabs=1.5e-8` and `epsrel=1.5e-8`. Per il prossimo esercizio, i valori di default sono sufficienti.

### Integrazione su un intervallo infinito

````python
def my_f(x):
    return 1/x/x


res, err = quad(my_f, 1, np.inf)
print(f"The numerical result is {res:f} (+-{err:g})")
````

### Integrazione di una funzione che dipende da parametri

Talvolta la funzione da integrare dipende da dei parametri. Per esempio la velocit&agrave; di un corpo che si muove in un campo gravitazionale costante &egrave; dato da $v(t) =  g t + v_0 $
dove $v_0$ &egrave; la velocit&agrave; a $t = 0$. Lo spostamento $y(t)$ pu&ograve; essere calcolato integrando questa relazione fra 0 e $t$<BR>
In questo caso la variabile su cui integrare deve essere al primo posto fra le variabili della funzione e i parametri addizionali vengono passati in una ntupla `args`:

````python
from scipy.integrate import quad

def vel(t, v0):
    g = -9.81
    return g*t + v0

v0 = 10  # m/s
tf = 3  # s

y, err = quad(vel, 0, tf, args=(v0,))
print(f"La posizione di un corpo che parte dall'origine con v0={v0} dopo {tf} secondi è {y} m")   
````

````python
def inv_square_root(x):
    return 1./np.sqrt(x)


res, err = quad(inv_square_root, 0, 1)
res, err
````

### Integrazione con metodo dei trapezi

</div>

Una routine elementare che illustra l'algoritmo per integrare  una funzione usando il metodo dei trapezi:

````python
# https://www.math.ubc.ca/~pwalls/math-python/integration/trapezoid-rule/

import numpy as np

def my_trapz(f, a, b, N=50):
    '''Approximate the integral of f(x) from a to b by the trapezoid rule.

    The trapezoid rule approximates the integral \int_a^b f(x) dx by the sum:
    (dx/2) \sum_{k=1}^N (f(x_k) + f(x_{k-1}))
    where x_k = a + k*dx and dx = (b - a)/N.

    Parameters
    ----------
    f : function
        Vectorized function of a single variable
    a , b : numbers
        Interval of integration [a,b]
    N : integer
        Number of subintervals of [a,b]

    Returns
    -------
    float
        Approximation of the integral of f(x) from a to b using the
        trapezoid rule with N subintervals of equal length.

    Examples
    --------
    >>> my_trapz(np.sin,0,np.pi/2,1000)
    0.9999997943832332
    '''
    x = np.linspace(a, b, N+1) # N+1 points make N subintervals
    y = f(x)
    y_right = y[1:] # right endpoints
    y_left = y[:-1] # left endpoints
    dx = (b - a)/N
    T = (dx/2) * np.sum(y_right + y_left)
    return T
````

````python
def vf(x):
    return np.exp(np.cos(-2 * x * np.pi))


res1 = my_trapz(vf, -2, 2)
print(f"The numerical result is {res1:f}")
````

Si pu&ograve; facilmente dimostrare che una equazione differenziale ordinaria di ordine `n`, cio&egrave; che coinvolge solo derivate fino all'ordine `n` della funzione incognita, pu&ograve; essere riscritta come un sistema di `n` equazioni differenziali di primo grado.
<BR>&nbsp;<BR>
Per risolvere  equazioni  differenziali ordinarie del tipo
$$\frac{\mathrm{d}y}{\mathrm{d}t}(t) = f(y,t)$$

con condizione iniziale $y(t_0)=y_0$, oppure un sistema di tali equazioni, si pu&ograve; usare la  funzione `odeint` di `scipy`. Ecco un esempio (`useodeint.py`) per determinare la soluzione 
$y(t)$, per $t$ nell'intervallo $[0,2]$
dell'equazione  differenziale:
$$\frac{\mathrm{d}y}{\mathrm{d}t}(t) = -2yt$$
con condizione iniziale
$$y(0)=1.$$

````python
import numpy as np
from scipy.integrate import odeint

def f(y, t):
    """this is the rhs of the ODE to integrate, i.e. dy/dt=f(y,t)"""
    return -2 * y * t

y0 = 1             # initial value
a = 0              # integration limits for t
b = 2

t = np.arange(a, b, 0.01)  # values of t for which we require the solution y(t)

y = odeint(f, y0, t)  # actual computation of y(t)

import matplotlib.pyplot as plt         # plotting of results
fig, ax = plt.subplots()
ax.plot(t, y)
ax.set_xlabel('t')
ax.set_ylabel('y(t)')
````

### Soluzione con il metodo di Eulero

Una routine elementare che illustra il metodo di Eulero per risolvere un'equazione differenziale in una variabile:

````python
# https://www.math.ubc.ca/~pwalls/math-python/differential-equations/first-order/

def odeEuler(f, y0, t):
    '''Approximate the solution of y'=f(y,t) by Euler's method.

    Parameters
    ----------
    f : function
        Right-hand side of the differential equation y'=f(t,y), y(t_0)=y_0
    y0 : number
        Initial value y(t0)=y0 where t0 is the entry at index 0 in the array t
    t : array
        1D NumPy array of t values where we approximate y values. Time step
        at each iteration is given by t[n+1] - t[n].

    Returns
    -------
    y : 1D NumPy array
        Approximation y[n] of the solution y(t_n) computed by Euler's method.
    '''
    y = np.zeros(len(t))
    y[0] = y0
    for n in range(0, len(t)-1):
        y[n+1] = y[n] + f(y[n], t[n])*(t[n+1] - t[n])
    return y
````

````python
y_E = odeEuler(f, y0, t)
````

````python
import matplotlib.pyplot as plt         # plotting of results
fig, ax = plt.subplots()
ax.plot(t, y, label='sol by odeint')
ax.plot(t, y_E, c='r', label='sol by odeEuler')
ax.set_xlabel('t')
ax.set_ylabel('y(t)')
ax.legend()
````

### Ricerca delle radici usando il metodo della bisezione 

L'algoritmo `bisect` &egrave; (i) robusto e (ii) concettualmente molto semplice (ma lento).

Supponiamo di dover calcolare le radici di $f(x)= x^3 − 2 x^2$. Questa funzione ha una radice (doppia) in $x = 0$ e un'altra fra $x = 1.5$ (dove $f(1.5) = − 1.125$) e $x = 3$ (dove $f(3) = 9$). &Egrave; facile vedere che questa altra radice si trova in $x = 2$. Ecco il programma che determina la radice numericamente:

````python
from scipy.optimize import bisect

def f(x):
    """returns f(x)=x^3-2x^2. Has roots at
    x=0 (double root) and x=2"""
    return x ** 3 - 2 * x ** 2

# main program starts here. Typically, the range in which to perform the search is determined from a plot.
x = bisect(f, 1.5, 3, xtol=1e-6)

print(f"The root x is approximately x={x:14.12g},\n"
      f"the error is less than 1e-6.")
print(f"The exact error is {2 - x:g}.")
````

Il metodo `bisect()` richiede obbligatoriamente tre argomenti: (i) la funzione `f(x)`, (ii) il limite inferiore `a` (che abbiamo scelto uguale a 1.5 nell'esempio) e (ii) il limite superiore `b` (scelto uguale a 3)  dell'intervallo in cui eseguire la procedura. Il parametro opzionale `xtol` determina l'errore massimo del metodo.

Uno dei presupposti del metodo di bisezione &egrave; che l'intervallo \[`a`, `b`\] sia scelto in modo tale che la funzione abbia in `a` segno opposto a quello che ha in `b` in modo che nell'intervallo, se la funzione &egrave; continua, cada almeno una radice.

### Ricerca delle radici usando la funzione `fsolve`

Un algoritmo per la ricerca delle radici che &egrave; (spesso) migliore (nel senso di “pi&ugrave; efficiente”) di quello di bisezione &egrave; codificato nella funzione `fsolve()` che funziona anche per problemi a pi&ugrave; dimensioni. Questo algoritmo richiede solamente un punto di partenza vicino a dove ci si aspetta che ci sia una radice (Non &egrave; per&ograve; detto che il metodo converga).

Ecco un esempio:

````python
from scipy.optimize import fsolve

def f(x):
    return x ** 3 - 2 * x ** 2


xstart = 3
x = fsolve(f, xstart)      # one root is at x=2.0
                           # search starts at x=3
                           # fsolve returns a numpy array

print(f"The root x is approximately x={x[0]:21.19g}")
print(f"The exact error is {2 - x[0]:g}.")
````

Il valore restituito da `fsolve` &egrave; un array di numpy di lunghezza $n$ per un problema di ricerca di radici con $n$ variabili. Nell'esempio precedente, $n = 1$.

#### Nota

## Ottimizzazione ovvero trovare minimi (e massimi)

Spesso &egrave; necessario trovare il minimo o il massimo di una particolare funzione `f(x)` dove `f` &egrave; una funzione scalare mentre `x` pu&ograve; essere un vettore. Applicazioni tipiche sono la minimizzazione di quantit&agrave; come il costo, il rischio o l'errore, oppure la massimizzazione della produttivit&agrave;, efficienza o profitto. Le routines di ottimizzazione tipicamente forniscono un metodo per minimizzare una funzione data: per massimizzare `f(x)` è sufficiente minimizzare `g(x)= − f(x)`.

Di seguito, un esempio che mostra (i) la definizione della funzione `f(x)` da minimizzare e (ii) la chiamata a `scipy.optimize.fmin` a cui vengono passati la funzione `f` da minimizzare e un valore iniziale `x0` da cui partire per la ricerca del minimo, e che restituisce il valore `xm` per cui `f(x)` ha un minimo. Tipicamente, la ricerca del minimo &egrave; locale, nel senso che l'algoritmo segue il gradiente (derivata multidimensionale) nel punto in cui si trova e non esplora tutto il range della funzione. 
Facciamo come prima cosa il grafico della funzione. Il plot suggerisce che i due minimi pi&ugrave; a destra possono essere cercati partendo dai due punti `x0 = 1.0` e `x1 = 2.0`. 

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fmin

def f(x):
    return np.cos(x) - 3 * np.exp(-(x - 0.2) ** 2)


# plot function
x = np.arange(-10, 10, 0.1)
y = f(x)

fig, ax = plt.subplots(figsize=(6, 4))
ax.set_xlabel('x')
ax.grid(visible=True, which='both')
ax.set_xlim(-5., 5.)
ax.set_ylim(-2.2, 0.5)

ax.plot(x, y, label=r'$\cos(x)-3e^{-(x-0.2)^2}$')
````

Troviamo i due minimi con due chiamate a `fmin`. Il risultato mostra come, in generale, partendo da punti diversi troviamo minimi diversi. &Egrave; anche possibile che esistano punti iniziali che non portano a nessun minimo. Rigeneriamo il plot della funzione, mostrando i punti iniziali delle due ricerche e i minimi ottenuti:

````python
# find minima of f(x), starting from 1.0 and 2.0 respectively
minimum1 = fmin(f, 1.0)
print("Start search at x=1., minimum is", minimum1)
minimum2 = fmin(f, 2.0)
print("Start search at x=2., minimum is", minimum2)

# plot function again showing the minima that have been found
x = np.arange(-10, 10, 0.1)
y = f(x)

fig, ax = plt.subplots(figsize=(6, 4))
ax.set_xlabel('x')
ax.grid(visible=True, which='both')
ax.set_xlim(-5.,5.)
ax.set_ylim(-2.2,0.5)

ax.plot(x, y, label=r'$\cos(x)-3e^{-(x-0.2)^2}$')
# add minimum1 to plot
ax.scatter(minimum1, f(minimum1), marker='v', c='r', label='minimum 1')
# add start1 to plot
ax.scatter(1.0, f(1.0), marker='o', c='r', label='start 1')

# add minimum2 to plot
ax.scatter(minimum2, f(minimum2), marker='v', c='g', label='minimum 2')
# add start2 to plot
ax.scatter(2.0, f(2.0), marker='o', c='g', label='start 2')

ax.legend(loc='lower left')
````

Come si vede, la chiamata a `fmin` produce anche dell'output addizionale che pu&ograve; essere utile per analizzare la procedura.

##### Valori ritornati da `fmin`

Si noti che la funzione `fmin` restituisce un numpy `array` che – nel caso precedente – contiene un solo numero dal momento che abbiamo una sola variabile (qui *x*) da variare. In generale, `fmin` pu&ograve; essre usata per trovare il minimo in uno spazio pluridimensionale. In questo caso, il numpy array contiene le coordinate del punto che minimizza la funzione obiettivo.

## Fit di un set di dati

Scipy fornisce una funzione piuttosto flessibile (basato sull'algoritmo di Levenburg-Marquardt), `scipy.optimize.curve_fit`, per interpolare un set di dati. L'assunzione &egrave; che vengano dati un set di punti
$x_1, x_2,\cdots,x_N$, i corrispondenti valori $y_i$ e una dipendenza funzionale $y=f(x,\vec{p})$.
Si vuole determinare i parametri $\vec{p}=(p_1, p_2, \ldots,p_k)$ in modo che $r$, la somma degli scarti quadratici fra la curva e i dati, sia la pi&ugrave; piccola possibile:

$$r = \sum\limits_{i=1}^N \left(y_i - f(x_i, \vec{p})\right)^2$$

Per fare un esempio, il numero di atomi non decaduti in un campione di una sostanza radioattiva segue la legge
$$ N(t,N_0,\tau) = N_0\,\exp\left(-\frac{t}{\tau}\right)$$
dove $N_0$ e $\tau$ sono i parametri che vogliamo determinare dalla misura dei decadimenti.

Questo approccio &egrave; particolarmente utile quando i dati sono affetti da *rumore*: per ogni coppia $x_i,y_i$ &egrave; presente un termine di errore (ignoto) $\epsilon_i$, tale che $y_i=f(x_i,\vec{p})+\epsilon_i$.

Un esempio per chiarire: assumiamo di avere dei dati che sappiamo essere descritti dalla funzione:
$$f(x,\vec{p}) = a \exp(-b x) + c,$$
che dipende dai parametri $\vec{p}=(a,b,c)$, che devono essere determinati usando i dati.

````python
import numpy as np
from scipy.optimize import curve_fit


def f(x, a, b, c):
    """Fit function y=f(x,p) with parameters p=(a,b,c). """
    return a * np.exp(- b * x) + c

#create fake data
x = np.linspace(0, 4, 50)
y = f(x, a=2.5, b=1.3, c=0.5)
#add noise. Typically, noise is already present in the data
#and there is no need to add it. 
yi = y + 0.2 * np.random.normal(size=len(x))

#call curve fit function
popt, pcov = curve_fit(f, x, yi)
# extract fit parameters
af, bf, cf = popt
print(f"Optimal parameters are af={af:g}, bf={bf:g}, and cf={cf:g}.")

# best fit curve
yfitted = f(x, af, bf, cf)

#plot
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(x, yi, marker='o', label=r'data $y_i$')
ax.plot(x, yfitted, c='r', label=r'fit $f(x_i)$')
ax.set_xlabel('x')
ax.legend()
````

Si noti che nell'esempio precedente abbiamo definito la funzione da utilizzare per il fit $y = f(x,\vec{p})$ usando del codice Python. Quindi possiamo utilizzare una funzione (quasi) arbitraria nel metodo `curve_fit`.

La funzione `curve_fit` restituisce una tuple `popt, pcov`. Il primo elemento, `popt`, contiene l'ntupla dei parametri ottimali (OPTimal Parameters), cio&egrave; i parametri che minimizzano la somma degli scarti quadratici. Il secondo elemento, `pcov`, contiene la matrice di covarianza di tutti i parametri (Sperimentazioni di Fisica). La diagonale fornisce la varianza della stima dei parametri. La standard deviation sulla stima dei parametri pu&ograve; quindi essere calcolata come
```
perr = np.sqrt(np.diag(pcov)).
```
L'algoritmo di Levenburg-Marquardt richiede, per iniziare la sua procedura, una stima iniziale dei parametri. Se questa non viene fornita, come nell'esempio precedente, il valore “1.0“ viene assunto come stima di partenza.

Se l'algoritmo non riesce a fittare i dati con la funzione ipotizzata, &egrave; necessario fornire a `curve_fit` una stima migliore per i parametri iniziali. Nell'esempio precedente potremmo passare le nostre stime cambiando la linea

```python
popt, pcov = curve_fit(f, x, yi)
```

in

```python
popt, pcov = curve_fit(f, x, yi, p0=(2,1,0.6))
```
se avessimo ragione di ritenere che *a* = 2, *b* = 1 and *c* = 0.6 siano dei valori "ragionevoli". Una volta che la stima iniziale &egrave; "più o meno corretta" il fit funziona bene.

## 🎯 Il Problema Centrale
**Quando le soluzioni analitiche non esistono o sono troppo complesse**, usiamo metodi numerici per:
- Calcolare integrali "impossibili"
- Risolvere equazioni differenziali non lineari
- Trovare zeri di funzioni complicate
- Interpolare ed estrapolare da dati sperimentali

## 📐 I Cinque Pilastri del Calcolo Numerico

### 1. **Integrazione Numerica** (`scipy.integrate`)
**Concetto chiave**: Approssimare l'area sotto una curva suddividendola in piccoli pezzi

- **`quad(f, a, b)`**: Metodo generale per integrali definiti
  - Restituisce: (risultato, errore stimato)
  - Gestisce anche limiti infiniti: `quad(f, 0, np.inf)`
- **`trapezoid`, `simpson`**: Per dati discreti (array di punti)
- **Parametri extra**: Passati con `args=(param1, param2)`

⚠️ **Attenzione**: Sempre plottare la funzione prima di integrarla per identificare singolarit&agrave;!

### 2. **Equazioni Differenziali** (`odeint`)
**Concetto chiave**: Trasformare equazioni di ordine `n` in `n` equazioni del primo ordine

- **Input essenziali**:
  1. Funzione `f(y, t)` che definisce `dy/dt`
  2. Condizione iniziale `y0`
  3. Array di tempi `t` dove calcolare la soluzione
- **Output**: Array con i valori di y(t)

**Pattern fisico tipico**: Oscillatore armonico, decadimento radioattivo, circuiti RC

### 3. **Ricerca di Zeri** (`scipy.optimize`)
**Concetto chiave**: Trovare x tale che f(x) = 0

- **`bisect(f, a, b)`**: Robusto ma lento
  - Richiede f(a) e f(b) di segno opposto
  - Garantisce convergenza se f &egrave; continua
- **`fsolve(f, x0)`**: Veloce ma richiede buon punto iniziale
  - Pu&ograve; non convergere
  - Restituisce array numpy

🔍 **Strategia**: Plot → identificare intervalli → scegliere metodo

### 4. **Ottimizzazione** (Massimi e Minimi)
**Concetto chiave**: I minimi sono locali, dipendono dal punto di partenza

- **`fmin(f, x0)`**: Trova minimo locale partendo da x0
- **Per massimi**: Minimizzare `-f(x)`
- **Output**: Array con coordinate del minimo

⚡ **Ricorda**: Diversi punti iniziali → diversi minimi!

### 5. **Fit di Dati** (`curve_fit`)
**Concetto chiave**: Minimizzare la somma degli scarti quadratici

- **Input necessari**:
  1. Funzione modello `f(x, param1, param2, ...)`
  2. Dati `x, y`
  3. (Opzionale) Stima iniziale parametri `p0`
- **Output**: 
  - `popt`: parametri ottimali
  - `pcov`: matrice di covarianza (errori su parametri)

## 💡 Workflow Universale per Problemi Numerici

1. **VISUALIZZA** sempre prima di calcolare
2. **IDENTIFICA** singolarit&agrave;, intervalli, punti iniziali dal grafico
3. **SCEGLI** il metodo appropriato
4. **VERIFICA** il risultato con un plot
5. **CONTROLLA** l'errore stimato

## ⚠️ Trappole Comuni da Evitare

1. **Mai fidarsi ciecamente**: Un algoritmo pu&ograve; convergere a risultati sbagliati
2. **Singolarità nascoste**: `f(x) = sin(1/x)` vicino a `x=0` &egrave; un incubo numerico
3. **Condizioni iniziali sbagliate**: In ottimizzazione e ODE determinano tutto
4. **Dimenticare l'errore**: `quad` restituisce (risultato, **errore**)
5. **Scale inappropriate**: Lavorare con numeri molto grandi o piccoli richiede attenzione

## 🔬 Applicazioni Fisiche Immediate

- **Meccanica**: Traiettorie con resistenza dell'aria (ODE non lineari)
- **Ottica**: Minimi per principio di Fermat
- **Laboratorio**: Fit di dati sperimentali con modelli teorici
- **Quantistica**: Ricerca autovalori (zeri del determinante)

## 📊 Scegliere il Metodo Giusto

| Problema | Metodo Robusto | Metodo Veloce | Quando Usarlo |
|----------|---------------|---------------|---------------|
| Integrale | `quad` | `simpson` | quad per funzioni, simpson per dati |
| Zero | `bisect` | `fsolve` | bisect se conosci intervallo, fsolve se hai buona stima |
| Minimo | - | `fmin` | Sempre con multiple partenze |
| ODE | `odeint` | Eulero (custom) | odeint sempre, Eulero solo per didattica |
